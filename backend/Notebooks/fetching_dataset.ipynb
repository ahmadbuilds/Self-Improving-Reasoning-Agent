```python
#fixing my PC to use the correct datasets library instead of the one from object detection api
import os
import sys
import importlib

os.environ.pop("PYTHONPATH", None)
blocked_tokens = [
    "object detection api/models/research",
    "object detection api/models/research/slim",
]

normalized_path = []
for entry in sys.path:
    if isinstance(entry, list):
        normalized_path.extend(entry)
    elif isinstance(entry, str):
        normalized_path.append(entry)

sys.path = [
    p
    for p in normalized_path
    if not any(token in p.replace('\\', '/').lower() for token in blocked_tokens)
]

sys.modules.pop("datasets", None)
importlib.invalidate_caches()

from datasets import load_dataset
print("Using datasets from module:", load_dataset.__module__)
```
